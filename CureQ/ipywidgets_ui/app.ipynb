Pages available:
- Analyze file: run a single-file analysis with progress/abort
- Parameters: edit and save analysis parameters
- Batch processing: run multiple files, per-file progress
- View results: inspect spikes/bursts and whole-well activity from outputs
- Compress/Rechunk: optimize HDF5 files (LZF/GZIP)
- Plotting: generate combined boxplots or features-over-time PDFs
- Exclude electrodes: toggle electrodes and recalculate features

```python
# 1) Install optional dependencies (if not already installed)
# Note: In JupyterLite, pip installs won't work; skip this step in the browser.
import sys, subprocess, pkgutil, os
is_pyodide = any(k in sys.modules for k in ("pyodide", "js")) or bool(os.environ.get("JUPYTERLITE_RUNTIME"))

def ensure(pkg):
    if pkg not in {m.name for m in pkgutil.iter_modules()}:
        print(f"Installing {pkg}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg])

for p in ("ipywidgets", "voila"):
    try:
        if not is_pyodide:
            ensure(p)
        else:
            print(f"Running in JupyterLite/Pyodide; skipping install of {p}")
    except Exception as e:
        print(f"Warning: could not verify/install {p}: {e}")
```

```python
# 2) Launch the MEAlytics ipywidgets UI
# In JupyterLite/Pyodide, multiprocessing is not available; the app will disable it automatically.
from CureQ.ipywidgets_ui import CureQApp
app = CureQApp()
# Ensure 'use multiprocessing' is off in browser environments just in case
try:
    import sys
    if any(k in sys.modules for k in ("pyodide", "js")):
        app.parameters["use multiprocessing"] = False
except Exception:
    pass
app.show()
```
